```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
PROJECT_ROOT = Path(r"C:\Case Study Trees\HomeCreditStudy")

RAW_DIR = PROJECT_ROOT / "data" / "raw"
SPLIT_DIR = PROJECT_ROOT / "data" / "splits"

SPLIT_DIR.mkdir(parents=True, exist_ok=True)

print("Raw data folder:", RAW_DIR)
print("Split output folder:", SPLIT_DIR)
print("Raw folder exists:", RAW_DIR.exists())
```

```python
df = pd.read_csv(RAW_DIR/'application_train.csv')
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
print("Missing applicant IDs:", df["SK_ID_CURR"].isna().sum())
print("Duplicate applicant IDs:", df["SK_ID_CURR"].duplicated().sum())
print("Missing targets:", df["TARGET"].isna().sum())

print("\nTarget counts:")
print(df["TARGET"].value_counts(dropna=False))
```

```python
train_valid_df, test_df = train_test_split(
    df,
    test_size=0.15,
    random_state=42,
    stratify=df["TARGET"]
)

print("Training + validation rows:", len(train_valid_df))
print("Test rows:", len(test_df))
```

```python
train_df, valid_df = train_test_split(
    train_valid_df,
    test_size=0.15 / 0.85,
    random_state=42,
    stratify=train_valid_df["TARGET"]
)

print("Training rows:", len(train_df))
print("Validation rows:", len(valid_df))
print("Test rows:", len(test_df))
```

```python
print("\nDefault percentage:")
print("Training:", round(train_df["TARGET"].mean() * 100, 2))
print("Validation:", round(valid_df["TARGET"].mean() * 100, 2))
print("Test:", round(test_df["TARGET"].mean() * 100, 2))

assert len(train_df) + len(valid_df) + len(test_df) == len(df)
```

```python
train_df.to_csv(SPLIT_DIR / "train.csv", index=False)
valid_df.to_csv(SPLIT_DIR / "valid.csv", index=False)
test_df.to_csv(SPLIT_DIR / "test.csv", index=False)

print("Saved to:", SPLIT_DIR)
print("train.csv:", train_df.shape)
print("valid.csv:", valid_df.shape)
print("test.csv:", test_df.shape)
```
